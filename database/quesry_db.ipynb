```python
import sqlite3
import pandas as pd

# 1. Establish the backend database connection path
DB_PATH = r"C:\Users\Mackay Iria\DataMiningProjects\telecom-edge-dw\database\telecom_warehouse.db"
conn = sqlite3.connect(DB_PATH)

# 2. Write your Raw, Advanced SQL Query inside a Python string
sql_query = """
SELECT 
    invoice_date,
    region,
    -- Simple Aggregation: Total revenue collected on this specific day
    SUM(total_invoice_amount) AS daily_revenue,
    
    -- ADVANCED WINDOW FUNCTION: Computes a running cumulative total 
    -- that automatically resets for each distinct region across time
    SUM(SUM(total_invoice_amount)) OVER (
        PARTITION BY region 
        ORDER BY invoice_date
        ROWS BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW
    ) AS cumulative_regional_revenue,
    
    -- Analytics Metric: Counts how many clean invoices were processed today
    COUNT(invoice_id) AS daily_invoice_count

FROM analytics_cleaned_invoices
GROUP BY invoice_date, region
ORDER BY region, invoice_date;
"""

# 3. Use Pandas to execute the SQL query engine and display the grid layout
matrix_df = pd.read_sql_query(sql_query, conn)

# Close the database warehouse connection safely
conn.close()

# 4. Render the final analytical matrix output
print("📈 --- Backend Data Warehouse SQL Query Execution Results --- 📈")
matrix_df
```
